## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Based on our EDA, we formed 3 hypotheses to validate using statistical tests. These are aimed at understanding whether:

**1.Agent shift affects satisfaction.**

**2.Channel of support influences CSAT.**

**3.Agent experience impacts CSAT.**

Each hypothesis is tested using an appropriate statistical test with interpretation of the result.

### Hypothesis 1: Agent Shift vs CSAT Score

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** There is no significant difference in CSAT Scores across different Agent Shifts.

**Alternate Hypothesis (H₁):** Agent Shift has a significant effect on CSAT Score.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway

morning = df[df['Agent Shift'] == 'Morning']['CSAT Score']
afternoon = df[df['Agent Shift'] == 'Afternoon']['CSAT Score']
night = df[df['Agent Shift'] == 'Night']['CSAT Score']

f_stat, p_val = f_oneway(morning, afternoon, night)
print("ANOVA Test p-value:", p_val)
```

##### Which statistical test have you done to obtain P-Value?

**ANOVA**

Agent Shift is a categorical variable with 3 groups

CSAT Score is a continuous dependent variable

ANOVA (Analysis of Variance) is the best choice to test whether means of 3 or more groups are significantly different.

##### Why did you choose the specific statistical test?

If p_val < 0.05:

We reject H₀ → Agent Shift does affect CSAT Score.

### Hypothesis 2: Support Channel vs CSAT Score

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** CSAT Score distribution is the same across different support channels (Chat, Call, Email).

**Alternate Hypothesis (H₁):** CSAT Score distribution differs based on the channel.

#### 2. Perform an appropriate statistical test.

```python
chat = df[df['channel_name'] == 'Chat']['CSAT Score']
email = df[df['channel_name'] == 'Email']['CSAT Score']
call = df[df['channel_name'] == 'Call']['CSAT Score']

f_stat2, p_val2 = f_oneway(chat, email, call)
print("Channel vs CSAT Score ANOVA p-value:", p_val2)
```

##### Which statistical test have you done to obtain P-Value?

**ANOVA**

channel_name has 3 categories

CSAT Score is continuous

ANOVA helps determine if at least one channel has significantly different CSAT scores

##### Why did you choose the specific statistical test?

If p_val2 < 0.05:

There is evidence that support channel affects CSAT, and Flipkart should focus more on underperforming channels.



### Hypothesis 3: Tenure Bucket vs CSAT Score

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): Agent Tenure has no impact on CSAT Score.

Alternate Hypothesis (H₁): Agent Tenure affects CSAT Score.

#### 2. Perform an appropriate statistical test.

```python
from scipy.stats import kruskal

tenure_groups = [group['CSAT Score'].values for name, group in df.groupby('Tenure Bucket')]
stat3, p_val3 = kruskal(*tenure_groups)
print("Tenure Bucket Kruskal-Wallis p-value:", p_val3)
```

##### Which statistical test have you done to obtain P-Value?

**Kruskal-Wallis**

Tenure Bucket has multiple categories

CSAT Score is ordinal, and we can’t guarantee normality

Kruskal-Wallis is a non-parametric test ideal for this setup

##### Why did you choose the specific statistical test?

If p_val3 < 0.05:

We reject H₀ → Tenure impacts CSAT Score, and experienced agents likely perform better.

### ML Model 3 – XGBoost Classifier


#### 1. Model Explanation

XGBoost (Extreme Gradient Boosting) is a powerful ensemble model that builds trees sequentially to correct errors from previous trees. It uses gradient boosting technique and supports regularization to reduce overfitting.

Why XGBoost?

✅ Excellent for handling imbalanced classes

✅ Supports multiclass classification

✅ Has built-in feature importance

✅ Faster & more accurate than traditional models

#### 2. Model Implementation

```python
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report

# Encode y labels for XGBoost
le_y = LabelEncoder()
y_train_enc = le_y.fit_transform(y_train)
y_test_enc = le_y.transform(y_test)

xgb = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
xgb.fit(X_train, y_train_enc)
y_pred_xgb = xgb.predict(X_test)

# Convert prediction back to original labels
y_pred_xgb_labels = le_y.inverse_transform(y_pred_xgb)

print("XGBoost Classifier Report:")
print(classification_report(y_test, y_pred_xgb_labels, zero_division=0))
```

| Low          | 0.59      | 0.16   | 0.25     |
| **Accuracy** | —         | —      | **83%**  |


Interpretation & Business Impact

| Metric              | Meaning                                          | Business Value                               |
| ------------------- | ------------------------------------------------ | -------------------------------------------- |
| **Accuracy**        | General correctness                              | Appears good, but could hide class imbalance |
| **Recall (Low)**    | How many unhappy customers are correctly flagged | High business value: prevents churn          |
| **Precision (Low)** | How many flagged are truly unhappy               | Reduces false escalations                    |
| **F1-score**        | Balanced score (especially for "Low")            | Most useful to judge real-world performance  |


 Compared to Logistic Regression, XGBoost significantly improved Low CSAT recall and F1-score.


#### Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'max_depth': [3, 5, 10],
    'learning_rate': [0.01, 0.1, 0.2],
    'n_estimators': [100, 200],
    'subsample': [0.7, 1]
}

rs_xgb = RandomizedSearchCV(
    estimator=XGBClassifier(use_label_encoder=False, eval_metric='mlogloss'),
    param_distributions=param_grid,
    n_iter=10,
    scoring='f1_weighted',
    cv=3,
    n_jobs=-1,
    random_state=42
)

rs_xgb.fit(X_train, y_train_enc)
best_xgb = rs_xgb.best_estimator_
y_pred_best = best_xgb.predict(X_test)
y_pred_best_labels = le_y.inverse_transform(y_pred_best)
```
